## 파일 및 예외처리

- 파일의 개념
파일은보조기억장치(영구저장장치)에기록된논리적인정보단위이다.
- 파일안에저장된정보는바이트단위로순차적으로기록되어있다.
- 첫번째바이트의위치를오프셋(offset) 0라고하고1씩증가한다.
- 파일을읽고쓰려면먼저파일을열어야한다.
- 열린파일포인터는최초에파일을열때는0을가리키며접근한바이트수만큼증가한다.
- 파일을접근할때는열린파일포인터의위치부터읽거나쓴다.
- 파일의끝을EOF (End of File) 라고한다.

```python
infile = open("123.png", "rb")
outfile = open("kkk.png", "wb")

while True:
    copy_buffer = infile.read(1024)
    if not copy_buffer:
        break
    outfile.write(copy_buffer)

infile.close()
outfile.close()
print("123.png를 kkk.png로 복사하였습니다. ")
```

```python
counter = [0] *26
infile = open("mobydick.txt", "r")
ch = infile.read(1)
while ch != "" :
    ch = ch.upper()		
    if ch >= "A" and ch <= "Z" :
        i = ord(ch) - ord("A")
        counter[i] += 1
    ch = infile.read(1)
print(counter)
```

```python
key = 'abcdefghijklmnopqrstuvwxyz'

def encrypt(n, plaintext):
    result = ''

    for l in plaintext.lower():
        try:
            i = (key.index(l) + n) % 26
            result += key[i]
        except ValueError:
            result += l

    return result.lower()

def decrypt(n, ciphertext):
    result = ''

    for l in ciphertext:
        try:
            i = (key.index(l) - n) % 26
            result += key[i]
        except ValueError:
            result += l

    return result

n = 3
text = 'The language of truth is simple.'
encrypted = encrypt(n, text)
decrypted = decrypt(n, encrypted)
print ('평문: ' ,  text)
print ('암호문: ',  encrypted)
print ('복호문: ' , decrypted)
```

```python
filename = input("파일명을 입력하세요: ").strip()
infile = open(filename, "r") 

freqs = {}
 
for line in infile:
    for char in line.strip():		
        if char in freqs:		
            freqs[char] += 1		
        else:				
            freqs[char] = 1		
print(freqs)
infile.close()
```
