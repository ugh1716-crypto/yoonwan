# 내장함수,람다식,제너레이터,모듈 연습노트
- 내장함수: 파이썬이터프린트에는 항상 사용할수있는 많은 함수가 준비되어있다.이러한 함수가 내장함수이다
- 람다식: 이름은 없고 몸체만있는 함수이다, 람다식은 lambda 키워드로 만들어진다, 람다식은 딱 한번 사용하는 함수를 만드는데 사용된다.
- 제너레이터: 키워드yield를 사용하여서 함수로부터 이터레이터를 생성하는 하나의 방법이다.
- 모듈: 함수나변수 또는 클래스들을 모아놓은 파일이다.

```python
class Student:
    def __init__ (self, name, grade, number):
        self.name = name
        self.grade = grade
        self.number = number

    def __repr__(self):
        return repr((self.name, self.grade, self.number))
students = [
Student('홍길동', 3.9, 20160303),
Student('김철수', 3.0, 20160302),
Student('최자영', 4.3, 20160301),
]
print(sorted(students, key=lambda student: student.number))
```

```python
class Person(object):
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __repr__(self):
        return "<이름: %s, 나이: %s>" % (self.name, self.age)

def keyAge(person):
    return person.age

people = [Person("홍길동", 20), Person("김철수", 35), Person("최자영", 38)]
print(sorted(people, key=keyAge)) 
```

# 제너레이터

```python
class MyCounter(object):

    def __init__(self, low, high):
        self.current = low
        self.high = high
 
    def __iter__(self):
        return self
    def __len__(self):
            return self.high - self.low + 1
    def __next__(self):
 
        if self.current > self.high:
            raise StopIteration
        else:
            self.current += 1
            return self.current - 1
    c = MyCounter(1, 10)
    for i in c:
        print(i, end=" ")
```

# 모듈

```python
def fib(n):
    a, b = 0, 1
    while b < n:
        print(b, end='')
        a, b= b, a+b
    print()

def fib2(n):
    result = []
    a, b = 0, 1
    while b < n:
        result.append(b)
        a, b = b, a+b
    return result
```
